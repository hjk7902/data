```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.support.ui import WebDriverWait, Select
from selenium.webdriver.common.by import By
from selenium.webdriver.support import expected_conditions as EC
```

```python
chrome_options = Options()
chrome_options.add_experimental_option("detach", True)  # 브라우저 창 닫히지 않게 유지
chrome_options.add_argument("--start-maximized")  # 브라우저 창 최대화
driver = webdriver.Chrome(options=chrome_options)
driver.get("https://www.g2b.go.kr/")
actions = ActionChains(driver)
# 페이지 완전 로드될 때까지 대기 (document.readyState == 'complete' 체크)
WebDriverWait(driver, 10).until(
    lambda d: d.execute_script("return document.readyState") == "complete"
)
print("페이지 로드 완료!")
```

```python
# "닫기" 버튼을 모두 기다린 후 닫기
WebDriverWait(driver, 10).until(
    EC.presence_of_all_elements_located((By.XPATH, "//input[contains(@id, 'popupCnts_btnClose')]"))
)
print("팝업 로드 완료!")

# 💡 모든 팝업 닫기 버튼 강제 클릭
popup_close_buttons = driver.find_elements(By.XPATH, "//input[contains(@id, 'popupCnts_btnClose')]")

if popup_close_buttons:
    for button in popup_close_buttons:
        # JS로 강제 클릭 (클릭 방해 대비)
        driver.execute_script("arguments[0].click();", button)
        print("팝업 닫기 완료!")
else:
    print("팝업이 없습니다.")
```

```python
wait = WebDriverWait(driver, 5)
입찰메뉴 = wait.until(EC.element_to_be_clickable(
  (By.ID, 'mf_wfm_gnb_wfm_gnbMenu_genDepth1_1_btn_menuLvl1_span'))
)
# 입찰메뉴 = driver.find_element(by=By.ID, value='mf_wfm_gnb_wfm_gnbMenu_genDepth1_1_btn_menuLvl1_span')
actions.move_to_element(입찰메뉴).click().perform()
```

```python
입찰공고목록메뉴 = wait.until(
  EC.element_to_be_clickable(
    (By.ID, 'mf_wfm_gnb_wfm_gnbMenu_genDepth1_1_genDepth2_0_genDepth3_0_btn_menuLvl3_span')
  )
)
actions.move_to_element(입찰공고목록메뉴).click().perform()

상세조건버튼 = wait.until(
  EC.presence_of_element_located(
    (By.CSS_SELECTOR, 'div#mf_wfm_container_tacBidPbancLst_contents_tab2_body_grpShbox [id$="_btnSearchToggle"]')
  )
)
actions.move_to_element(상세조건버튼).click().perform()

업무구분_전체버튼 = wait.until(
  EC.element_to_be_clickable(
    (By.ID, 'mf_wfm_container_tacBidPbancLst_contents_tab2_body_chkBidPbancSrchTyCd_input_0')
  )
)
actions.move_to_element(업무구분_전체버튼).click().perform()

업무구분_공사버튼 = wait.until(
  EC.element_to_be_clickable(
    (By.ID, 'mf_wfm_container_tacBidPbancLst_contents_tab2_body_chkBidPbancSrchTyCd_input_4')
  )
)
actions.move_to_element(업무구분_공사버튼).click().perform()

공고기관버튼 = wait.until(
  EC.element_to_be_clickable(
    (By.ID, 'mf_wfm_container_tacBidPbancLst_contents_tab2_body_untyGrpGb1_input_0')
  )
)
actions.move_to_element(공고기관버튼).click().perform()

기관명입력필드 = wait.until(
  EC.element_to_be_clickable(
    (By.ID, 'mf_wfm_container_tacBidPbancLst_contents_tab2_body_ibxDmstNm1')
  )
)
actions.move_to_element(기관명입력필드).click().perform()
기관명입력필드.send_keys('과학기술정보통신부')

참가제한지역 = wait.until(
  EC.element_to_be_clickable(
    (By.ID, 'mf_wfm_container_tacBidPbancLst_contents_tab2_body_sbxOdnLmtLgdngCd')
  )
)
참가제한지역선택 = Select(참가제한지역)
참가제한지역선택.select_by_visible_text('서울특별시')

검색버튼 = wait.until(
  EC.element_to_be_clickable(
    (By.ID, 'mf_wfm_container_tacBidPbancLst_contents_tab2_body_btnS0004')
  )
)
actions.move_to_element(검색버튼).click().perform()

상세조건버튼.click()

입찰공고목록결과테이블 = wait.until(
  EC.element_to_be_clickable(
    (By.ID, 'mf_wfm_container_tacBidPbancLst_contents_tab2_body_gridView1_body_table')
  )
)

입찰공고목록리스트 = 입찰공고목록결과테이블.find_elements(
  By.XPATH, 
  '//*[@id="mf_wfm_container_tacBidPbancLst_contents_tab2_body_gridView1_body_tbody"]//tr[@style and not(@style!="")]' # style 속성이 있는 tr 태그만 선택
)
print("조회된 공고 수:", len(입찰공고목록리스트))
```

```python
print("====================조회된 공고명====================")
for 입찰공고 in 입찰공고목록리스트:
    공고명요소 = 입찰공고.find_element(By.CSS_SELECTOR, 'td:nth-of-type(7)')
    공고명 = 공고명요소.text
    if(공고명):
        print(공고명)
```
