# Multiplot Grids

## FacetGrid

```python
import seaborn as sns
sns.set()
iris = sns.load_dataset("iris")

g = sns.FacetGrid(iris, col="species", hue="species")
g.map(plt.hist, "sepal_length")
g.set_axis_labels(y_var="Count")
```

```python
g = sns.FacetGrid(iris, col="species")
g.map(sns.scatterplot, 'petal_width', 'petal_length', 
      hue=iris.species, size=iris.sepal_length)
```

```python
import seaborn as sns
titanic = sns.load_dataset("titanic")

g = sns.FacetGrid(titanic, col="survived", row="sex")
g.map(plt.hist,"age")
```

## PariGrid

```python
g = sns.PairGrid(iris)
g.map(plt.scatter)
```

```python
g = sns.PairGrid(iris)
g.map_diag(sns.kdeplot)
g.map_lower(sns.scatterplot)
g.map_upper(sns.regplot)
```

## JointGrid

```python
import seaborn as sns
iris = sns.load_dataset("iris")

g = sns.JointGrid(x="petal_length", y="petal_width", data=iris)
g.plot_joint(sns.scatterplot)
g.plot_marginals(sns.histplot)
```

# 그래프 커스터마이징

```python
import seaborn as sns
iris = sns.load_dataset("iris")

import matplotlib.pyplot as plt
plt.boxplot(iris["sepal_width"])
plt.show()
```

```python
import numpy as np
import pandas as pd
Q1,Q3 = np.percentile(iris["sepal_width"], [25, 75])
IQR = Q3 - Q1
outlier_cond = (iris["sepal_width"] < Q1-1.5*IQR) | (iris["sepal_width"] > Q3+1.5*IQR)
iris.loc[outlier_cond, :]
```

* 연속된 날짜를 그래프에 표시하려면

```python
y = [29, 30, 31, 51, 104, 204, 433, 602, 833, 977, 1261, 1766, 2337, 3150]
 
from datetime import date, timedelta
sdate = date(2020, 2, 16)   # start date
edate = sdate + timedelta(days=len(y))	#date(2020, 9, 15)   # end date
date_list = [sdate + timedelta(days=x) for x in range((edate - sdate).days)]
x = [date_obj.strftime('%m-%d') for date_obj in date_list]
 
import matplotlib.pyplot as plt
import seaborn as sns
 
sns.scatterplot(x=x, y=y)
plt.xticks(rotation=90)
plt.show()
```
